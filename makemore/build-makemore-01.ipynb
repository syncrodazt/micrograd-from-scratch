```python
word_set = sorted(set(''.join(words)))
stoi = {s:i for s, i in zip(word_set, range(1, 27))}
stoi['.'] = 0
itos = {i:s for s, i in stoi.items()}
```

```python
import torch
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
N = torch.zeros(27, 27).int()
```

```python
for w in words[:1]:
    tokens = "." + w + "."
    for i in range(len(tokens)):
        ch1, ch2 = tokens[i], tokens[i+1]
        print(ch1, ch2)
```

```python
for w in words:
    tokens = "." + w + "."
    for i in range(1, len(tokens)):
        ch1, ch2 = tokens[i-1], tokens[i]
        ix1 = stoi[ch1]
        ix2 = stoi[ch2]
        N[ix1, ix2] += 1
```

```python
plt.imshow(N, cmap="Blues")
```
