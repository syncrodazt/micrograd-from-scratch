```python
from math import comb, exp, pi, sqrt
n,k=10,7
print('เหรียญ: โยน 10 ครั้ง ออกหัว 7 ครั้ง')
print()
print('PROBABILITY - ตรึง p = 0.5 แล้วไล่ *ข้อมูล* (จำนวนหัว 0..10)')
tot=0
for h in range(11):
    v=comb(n,h)*0.5**n; tot+=v
    print('   หัว %2d ครั้ง  ->  %.4f' % (h,v))
print('   รวม        ->  %.4f   <-- ได้ 1 พอดี' % tot)
print()
print('LIKELIHOOD - ตรึงข้อมูล (หัว 7) แล้วไล่ *พารามิเตอร์* p')
s=0
for p in [0.0,0.1,0.3,0.5,0.7,0.9,1.0]:
    v=comb(n,k)*p**k*(1-p)**(n-k); s+=v
    m='   <-- สูงสุด = 7/10' if abs(p-0.7)<1e-9 else ''
    print('   p = %.1f      ->  %.4f%s' % (p,v,m))
print('   รวม        ->  %.4f   <-- ไม่ได้ 1 และไม่ได้แปลว่าอะไร' % s)
print()
print('LIKELIHOOD เกิน 1 ได้จริงไหม? - ข้อมูลต่อเนื่อง, normal density ที่ยอด')
for sd in [1.0,0.5,0.1,0.01]:
    print('   sigma = %.2f  ->  likelihood ที่ x=0 = %.4f' % (sd, 1/(sd*sqrt(2*pi))))
```
