```python
def make_counter():
    n = 0
    def bump():
        nonlocal n
        n += 1
        return n
    return bump          # ← make_counter จบตรงนี้ n ควรจะตายแล้ว

c1 = make_counter()
c2 = make_counter()

print(c1(), c1(), c1())   # 1 2 3
print(c2())               # 1   ← ไม่ใช่ 4
```
